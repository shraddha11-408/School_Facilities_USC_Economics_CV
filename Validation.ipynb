```python
import pandas as pd
from google.colab import files

# Upload measurements.csv
uploaded = files.upload()
filename = next(iter(uploaded))

measurements = pd.read_csv(
    filename,
    dtype={"school_id": "string"},
)

measurements["school_id"] = (
    measurements["school_id"]
    .astype("string")
    .str.strip()
)

# Sort first so the original row order does not affect selection
measurements = measurements.sort_values(
    "school_id"
).reset_index(drop=True)

# Fixed selection: always produces the same eight schools
random_8_schools = measurements.sample(
    n=8,
    random_state=42,
).sort_values("school_id").reset_index(drop=True)

display_columns = [
    column
    for column in [
        "school_id",
        "school_name",
        "district",
        "city",
        "state",
        "level",
    ]
    if column in random_8_schools.columns
]

display(random_8_schools[display_columns])
```

```python
import pandas as pd

number_of_verified_schools = 8

wrong_predictions = {
    "solar_panels": {
        "hand_labelled": 1,
        "groq_qwen": 2,
    },
    "portable_classrooms": {
        "hand_labelled": 0,
        "groq_qwen": 0,
    },
    "running_track": {
        "hand_labelled": 0,
        "groq_qwen": 0,
    },
    "full_sized_field": {
        "hand_labelled": 1,
        "groq_qwen": 4,
    },
    "hard_courts": {
        "hand_labelled": 0,
        "groq_qwen": 3,
    },
    "pool": {
        "hand_labelled": 0,
        "groq_qwen": 0,
    },
}

results = []

for field, errors in wrong_predictions.items():
    results.append({
        "field": field,
        "verified_schools": number_of_verified_schools,
        "hand_wrong": errors["hand_labelled"],
        "hand_error_rate_percent": (
            errors["hand_labelled"]
            / number_of_verified_schools
            * 100
        ),
        "groq_qwen_wrong": errors["groq_qwen"],
        "groq_qwen_error_rate_percent": (
            errors["groq_qwen"]
            / number_of_verified_schools
            * 100
        ),
    })

error_rates = pd.DataFrame(results)

error_rates[
    [
        "hand_error_rate_percent",
        "groq_qwen_error_rate_percent",
    ]
] = error_rates[
    [
        "hand_error_rate_percent",
        "groq_qwen_error_rate_percent",
    ]
].round(1)

display(error_rates)
```
