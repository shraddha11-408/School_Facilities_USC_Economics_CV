```python
import pandas as pd
from google.colab import files

# ---------------------------------------------------------
# 1. Upload schools_sample.csv
# ---------------------------------------------------------
print("Upload schools_sample.csv")
uploaded = files.upload()

input_filename = next(iter(uploaded))

# Read everything as text so NCES school IDs retain leading zeros
schools = pd.read_csv(input_filename, dtype=str)

# Standardize input column names:
# "school id" becomes "school_id", "School Name" becomes "school_name", etc.
schools.columns = (
    schools.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

if "school_id" not in schools.columns:
    raise ValueError(
        f"Could not find the school ID column. Columns found: {schools.columns.tolist()}"
    )

# Check that each school ID appears only once
if schools["school_id"].duplicated().any():
    duplicate_ids = schools.loc[
        schools["school_id"].duplicated(keep=False), "school_id"
    ].tolist()
    raise ValueError(f"Duplicate school IDs found: {duplicate_ids}")

# ---------------------------------------------------------
# 2. Copy useful identifying columns from the source file
# ---------------------------------------------------------
identifier_columns = [
    "school_id",
    "school_name",
    "district",
    "city",
    "state",
    "level",
    "enrollment_2024",
    "latitude",
    "longitude",
]

# Only copy columns that actually exist
identifier_columns = [
    column for column in identifier_columns if column in schools.columns
]

measurements = schools[identifier_columns].copy()

# ---------------------------------------------------------
# 3. Add blank measurement and confidence columns
# ---------------------------------------------------------
measurement_columns = [
    # A. Rooftop solar
    "solar_present",
    "solar_present_confidence",
    "solar_area_m2",
    "solar_area_m2_confidence",

    # B. Portable/modular classrooms
    "portable_classrooms_count",
    "portable_classrooms_count_confidence",

    # C. Perimeter fencing
    "perimeter_fencing",
    "perimeter_fencing_confidence",
    "dominant_fence_type",
    "dominant_fence_type_confidence",

    # D. Athletic facilities
    "running_track",
    "running_track_confidence",
    "full_size_sports_fields_count",
    "full_size_sports_fields_count_confidence",
    "hard_courts_count",
    "hard_courts_count_confidence",
    "pool_present",
    "pool_present_confidence",

    # Imagery information and notes
    "imagery_source",
    "imagery_date_or_vintage",
    "review_notes",
]

for column in measurement_columns:
    measurements[column] = ""

# ---------------------------------------------------------
# 4. Save and download measurements.csv
# ---------------------------------------------------------
output_filename = "measurements.csv"
measurements.to_csv(output_filename, index=False)

print(f"\nCreated {output_filename}")
print(f"Rows: {len(measurements)}")
print(f"Columns: {len(measurements.columns)}")
print("\nPreview:")
display(measurements.head())

files.download(output_filename)
```
